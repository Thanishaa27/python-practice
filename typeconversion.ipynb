```python
#int conversion
#eg1:float to int
a=10.20
type(a)
b=int(a)
print(b,type(b))
```

```python
#eg2: complex to int
x=17.56+20j
y=int(x.real)
z=int(x.imag)
print(y)
print(z)
#a=int(X)error
```

```python
#eg3:str  to int
x='500'
t=int(x)
print(t,type(t))
```

```python
#eg4:bool to int
x=False
y=int(x)
print(y,type(y))
```
