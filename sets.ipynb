```python
#set of squares of numbers from 0 to 4
squares={x**2 for x in range(5)}
print(squares)
```

```python
#creating a new set of squared values
num=[1,2,3,4,5]
squared_no={x**2 for x in num}
print(squared_no)
```

```python
#create a set of squares,but only for even numbers
nums=[1,2,3,4,5]
even_num={x for x in nums if x%2==0}
print(even_num)
```

```python
#using a set comprehension with sstrings
#say the fruit name without vowel!
words=['apple','banana','cherry','lyn']
vowels={'a','e','i','o','u'}
vowel_words={word for word in words if any(letter in vowels for letter in word)}
print(vowel_words)
```

```python
#eliminating duplicates
duplicates=[1,2,3,4,5,1,6]
unique_squares={x**2 for x in duplicates}
print(unique_squares)
```

```python
#set comprehension with string manipulation,create a set of uppercase letters from a string
text="hello world"
uppercase_letters={char.upper() for char in text if char.isalpha()}
print(uppercase_letters)
```
