```python
#empty dictionary
dic={}
print(dic)
```

```python
#dictionary with integer keys
dic={1:"Focus",2:"academy",3:"for",4:"career",5:"Enhancement"}
print(dic)
```

```python
#dictionary with mixed keys
dic={1:(2,4,5),"Name":"Face",4:[5,7]}
print(dic)
```

```python
#dic with dict()
dic=dict({1:[2,4,5],"Name":"FACE",4:[5,7]})
print(dic)
```

```python
#creating a dictionary with pair ele
dic=dict([(1,2),("Name","FACE"),(4,5)])
print(dic)
```

```python
#creating a nested dictionary
dic={1:"FACE",2:4,3:{5:"HELLO",6:9.5}}
print(dic)
```

```python
#adding ele to a dictionary
dic={}
dic[0]="Apple"
dic[1]="Hard"
dic[3]="work"
print(dic)
dic['new_set']=1,5,8
print(dic)
```

```python
dic={0: 'Apple', 1: 'Hard', 3: 'work', 'new_set': (1, 5, 8)}
print(dic["new_set"])
print(dic[1])
```

```python
#access elements using get()
dic={0: 'Apple', 1: 'Hard', 3: 'work', 'new_set': (1, 5, 8)}
print(dic.get("new_set"))
print(dic.get(1))
```

```python
#removing ele
dic={0: 'Apple', 1: 'Hard', 3: 'work', 'new_set': (1, 5, 8)}
dic.pop("new_set")
print(dic)
```

```python
dic={0: 'Apple', 1: 'Hard', 3: 'work', 'new_set': (1, 5, 8)}
print(frozenset(dic))
```
